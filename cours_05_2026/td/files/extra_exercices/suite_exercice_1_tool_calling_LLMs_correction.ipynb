# 11. Exercices supplmentaires

## Exercice A — Créer un nouveau tool

Créez un tool `convert_km_to_miles(km: float)`.

Contraintes :

- la fonction retourne un dictionnaire ;
- le schéma JSON expose un paramètre `km` de type `number` ;
- le modèle doit pouvoir répondre à :

> Convertis 12 kilomètres en miles.

---

## Exercice B — Ajouter un enum

Créez un tool `convert_temperature(value: float, from_unit: str, to_unit: str)`.

Contraintes :

- `from_unit` doit être `celsius` ou `fahrenheit` ;
- `to_unit` doit être `celsius` ou `fahrenheit` ;
- utilisez `enum` dans le JSON Schema.

---

## Exercice C — Tool avec erreur métier

Créez un tool `get_product_price(product_name: str)` qui retourne une erreur si le produit n'existe pas.

Le modèle doit expliquer correctement l'erreur à l'utilisateur.

---

## Exercice D — Tool dangereux

Imaginez un tool `delete_customer(customer_id: str)`.

Ne l'exécutez pas réellement. Écrivez seulement :

1. son schéma ;
2. la logique de confirmation humaine ;
3. les informations à logger.
